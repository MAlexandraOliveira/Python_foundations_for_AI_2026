## 2. How AI Apps Work Behind the Scenes

When you type a question into an AI assistant, what actually happens?

```
👤 User types a question
         │
         ▼
🐍 Python receives the input
         │
         ▼
🔧 Python prepares the prompt
   (cleans text, adds context, formats the request)
         │
         ▼
🌐 Python calls the AI model via API
   (sends the prepared prompt to Gemini, GPT, Claude...)
         │
         ▼
📨 Python receives the model's response
         │
         ▼
✅ Python processes and displays the answer
         │
         ▼
👤 User sees the response
```

**Python is the glue** that connects every part of this pipeline.

The AI model itself is just one component. Python handles everything around it:
reading inputs, cleaning text, calling APIs, formatting outputs, storing results.

## 5. Variables, Program Flow & Data Types

### Variables

A **variable** is a named container that stores a value. You create one by writing `name = value`.

| ✅ Valid | ❌ Invalid | Why |
|----------|-----------|-----|
| `model_name` | `model-name` | no hyphens |
| `modelName` | `2nd_model` | cannot start with a number |
| `MODEL` | `model name` | no spaces |
| `model2` | `class` | reserved Python word |

The convention in Python is `lower_case_with_underscores`.

````python
model_name = "FALC-ON AI"
print(model_name)
````

````python
# A variable can be updated - the old value is replaced
model_name = "FALC-ON AI v2"
print(model_name)
````

### Program flow — one line at a time

Now that a variable can change, the **order of the lines** matters. Python runs your code **line by line, top to bottom**, like a recipe:



The second `print()` shows a different number because the line above it changed `tokens`.

````python
tokens = 100
print(tokens)             # shows 100
````

````python
tokens = tokens + 50      # this line changes tokens
print(tokens)             # shows 150 - the order of the lines matters
````

### Data types

Every value in Python has a **type**: it tells Python what kind of data it is and what you can do with it.

A variable can hold a value of **any** type — text, a number, `True`/`False`, a list. You never say which type: Python works it out from the value you store.

Today we use five:

| Type | Keyword | Example | Common use in AI |
|------|---------|---------|------------------|
| Text | `str` | `"Hello"` | Prompts, responses, labels |
| Whole number | `int` | `42` | Token counts, loop counters |
| Decimal | `float` | `0.95` | Confidence scores, temperatures |
| True / False | `bool` | `True` | Flags, conditions, filters |
| A sequence | `list` | `["a", "b"]` | Several values at once |

> 💡 There is a sixth type you will use constantly in AI — the **dictionary** (`dict`). It is how JSON data and API responses are structured. It is the main topic of **Class 2**.

````python
# One AI request, stored across several variables
prompt  = "What is machine learning?"    # str   - text
tokens  = 128                            # int   - a whole number
score   = 0.92                           # float - a decimal number
is_safe = True                           # bool  - True or False
topics  = ["AI", "Python", "Data"]       # list  - several values

print(prompt, tokens, score, is_safe, topics)
````

### `type()` — which type is this?

`type()` tells you the type of any value.

````python
print(type(prompt))     # type: str
print(type(tokens))     # type: int
print(type(score))      # type: float
print(type(is_safe))    # type: bool
print(type(topics))     # type: list
````

### A variable can change type

The type belongs to the **value**, not to the variable. Store a different kind of value, and the variable's type changes with it.

````python
answer = 42
print(type(answer))     # type: int

answer = "forty-two"    # the same variable, now holding text
print(type(answer))     # type: str
````

### `len()` — how many items?

`len()` tells you how many items a list holds.

````python
print(len(topics))      # 3 - how many items in the list
````

## 6. Casting and `input()`

### Casting — changing a value's type

Sometimes a number arrives as text — for example `"42"` read from a file or typed by a user.
You cannot do arithmetic with text, so you **cast** it: `int()` for whole numbers, `float()` for decimals, `str()` for text.

````python
x = "42"            # this is TEXT, not a number
print(type(x))      # type: str
````

````python
x = int(x)          # cast it to a whole number
print(type(x))      # type: int
print(x + 1)        # now arithmetic works: 43
````

````python
# float() for numbers with decimals
temperature = float("0.7")
print(temperature * 2)          # 1.4
````

> ⚠️ **Text is glued, not added.** `"10" + "5"` gives `"105"`, not `15`. Forgetting to cast is one of the most common beginner bugs.

````python
print("10" + "5")             # 105 - two pieces of text glued together
print(int("10") + int("5"))   # 15  - two numbers added
````

### `input()` — asking the user

`input()` pauses the program and waits for someone to type. This is how a chatbot receives a question.

Whatever the user types comes back as **text**, always — so if you need a number, you must cast it.

````python
name = input("What's your name? ")
print("Hello,", name)
````

````python
typed = input("How many questions will you ask today? ")
questions = int(typed)                   # text -> whole number
print("Tokens needed:", questions * 150)
````

## 7. Operators

### Arithmetic

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `3 + 2` | `5` |
| `-` | Subtraction | `3 - 2` | `1` |
| `*` | Multiplication | `3 * 2` | `6` |
| `/` | Division | `7 / 2` | `3.5` |
| `**` | Power | `2 ** 3` | `8` |

````python
total_tokens = 1200
used_tokens  = 850

remaining = total_tokens - used_tokens
print("Remaining tokens:", remaining)
````

````python
pct_used = used_tokens / total_tokens * 100
print("Percentage used:", pct_used)
````

### `round()` — fewer decimals

`round(value, 1)` rounds a number to 1 decimal place.

````python
print(round(pct_used, 1))     # 70.8 - much easier to read
````

### Assignment shortcuts

`+=` and `-=` update a variable using its own value. You will use `+=` constantly to count things inside a loop.

| Shortcut | Same as |
|----------|---------|
| `x += 1` | `x = x + 1` |
| `x -= 1` | `x = x - 1` |

````python
question_count = 0
question_count += 1           # same as: question_count = question_count + 1
question_count += 1
print("Questions asked:", question_count)    # 2
````

### Comparison — always gives `True` or `False`

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal | `x == 5` |
| `!=` | Not equal | `x != 5` |
| `>` | Greater than | `x > 5` |
| `<` | Less than | `x < 5` |
| `>=` | Greater or equal | `x >= 5` |
| `<=` | Less or equal | `x <= 5` |

> ⚠️ `=` **stores** a value in a variable. `==` **compares** two values. Mixing them up is a classic bug.

````python
print(remaining > 0)                  # True
print(remaining < 100)                # False
print(used_tokens == total_tokens)    # False - not at the limit yet
````

### Logical — combining conditions

| Operator | Meaning | Example |
|----------|---------|--------|
| `and` | Both must be True | `x > 0 and x < 10` |
| `or` | At least one True | `x < 0 or x > 10` |
| `not` | Reverses the result | `not (x == 5)` |

````python
budget_ok   = remaining > 100
model_ready = True

print("Can run AI call:", budget_ok and model_ready)   # True
print(not model_ready)                                 # False
````

### Membership — is this word in the text? `in` and `not in`

`in` checks whether a word appears inside a text. It is how our stand-in model will spot keywords in the capstone. (In Class 2, you will use the same `in` with lists.)

| Operator | Meaning | Example |
|----------|---------|--------|
| `in` | Is this word inside the text? | `"refund" in message` |
| `not in` | The opposite | `"login" not in message` |

````python
message = "I want a refund for my order"

print("refund" in message)      # True
print("login" in message)       # False
print("login" not in message)   # True
````

## 8. f-strings — putting values inside text

Until now, to show text and a variable together, we separated them with **commas**: `print("Tokens used:", tokens)`. That works, but `print()` puts a space between every piece, and the sentence is chopped into parts.

````python
user = "Ana"
print("Hello", user, "!")      # Hello Ana ! - the commas add a space before the !
````

An **f-string** keeps the sentence in one piece: it is a text with an `f` before the opening quote, and anything you write inside `{}` is replaced by its value — a variable, or even a calculation.

> 💡 Use f-strings from now on. They are the standard way to build output in modern Python, and you will see them in every remaining class.

````python
print(f"Hello {user}!")        # Hello Ana! - the value goes exactly where you put it
````

````python
# Several values in the same text
asked = 3
print(f"{user} asked {asked} questions")
````

````python
# A calculation inside the {}
print(f"Tokens left: {total_tokens - used_tokens}")
````

## 9. Making Decisions — `if` / `elif` / `else`

`if` lets your program choose what to do. `elif` ("else if") tries another condition, and `else` catches everything left over.
Python checks the conditions **in order** and stops at the first one that is `True`.

This is how AI systems route requests. For example, deciding what to do after a customer rates a chatbot conversation from 1 to 5:



> ⚠️ **Indentation is part of the language.** The lines inside an `if` must be indented (4 spaces, or one Tab). Python uses that indentation to know which lines belong to the block — there are no brackets.

````python
score = 2                  # the customer's rating, from 1 to 5

if score >= 4:
    print("Happy customer - no action needed")
elif score == 3:
    print("Neutral - send a follow-up email")
else:
    print("Unhappy customer - escalate to a human agent")
````

### Combining conditions

A condition can use `and`, `or` and `in`, like any other comparison. Here a message is routed by its keywords.

````python
message = "my card was charged twice"

if "refund" in message or "charged" in message:
    print("Route to the billing team")
else:
    print("Route to general support")
````

## 10. Repeating Work — `while`, `for`, `break`, `continue`

Doing the same thing to 1,000 messages is what computers are for. A **loop** repeats work. Python has two.

### `while` — repeat as long as something is true

Use `while` when you don't know in advance how many times you will go round. The arrow back to the test is the important part:



> ⚠️ Something inside the loop must eventually make the condition `False`, or it runs forever. In Colab, press the ■ stop button if that happens.

````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1        # without this line, the loop would never end

print("Request sent!")
````

### `for` — do something once per item

Use `for` when you know what you are working through, such as every item in a list.
You don't manage a counter — Python does it for you.

````python
# One turn per support ticket
tickets = ["billing issue", "login problem", "refund request"]

for ticket in tickets:
    print(f"Processing ticket: {ticket}")
````

### `range()` — repeat a fixed number of times

`range(n)` generates the numbers from `0` up to `n - 1`: `range(3)` gives `0, 1, 2`.

````python
for i in range(3):
    print(i)
````

### `break` — leave the loop

`break` leaves the loop **completely**, immediately. Use it when you got what you were waiting for.

`while True:` is a loop that would repeat forever: `break` is its only way out.
It is exactly how a chatbot keeps talking until the user says goodbye.

````python
# while True + break - retry an AI call until it succeeds
attempt = 0

while True:
    attempt += 1
    print(f"Calling the model - attempt {attempt}")
    if attempt == 3:
        print("Response received!")
        break             # leaves the loop completely
````

### `continue` — skip to the next turn

`continue` skips the rest of **this one turn** and goes straight to the next item. Use it when an item is not relevant.

````python
# Skip empty messages, send the rest
messages = ["What is AI?", "", "How do I learn Python?"]

for message in messages:
    if message == "":
        print("Empty message - skipping")
        continue          # jump straight to the next message
    print(f"Sending to the model: {message}")
````

> 🤖 **AI Connection**
>
> Control flow is the backbone of every AI pipeline. `if/else` routes requests to the right model or handler. `for` loops process batches of documents or messages. `while` loops keep a conversation going until the user says goodbye.

## 11. 🔗 Capstone — FALC-ON AI, Our First AI Pipeline

Remember the pipeline from **Section 2**? We now build it with today's Python.
**FALC-ON AI** answers questions about this course, following the same boxes:

1. 👤 **Receive** the user's question
2. 🔧 **Prepare** the prompt
3. 🌐 **Call the model**
4. ✅ **Display** the answer
5. 🔁 **Repeat**, until the user types `bye`

We don't call a real AI model yet: that comes in **Class 9**. Today, a few `if` / `elif` rules play the role of the model.
Later, only Step 3 changes — the rest of the pipeline stays exactly the same.

### Step 1 — 👤 Receive the question

Run the cell and type `how many classes?`

````python
question = input("You: ")
print(f"Received: {question}")
````

### Step 2 — 🔧 Prepare the prompt

A **prompt** is the text we send to the model. We wrap the user's question in instructions with an f-string, so the model knows how to answer.

````python
prompt = f"You are a course assistant. Answer in one short sentence: {question}"
print(prompt)
````

### Step 3 — 🌐 Call the model (a stand-in, for today)

A real model would read the prompt and write an answer. Our stand-in just looks for a keyword in the question, and picks a prepared response.

> 💡 The stand-in only finds a keyword written exactly the same way: `"class"` is not the same as `"Class"`. Type your questions in lowercase.

````python
if "class" in question:
    response = "The course has 10 classes."
elif "teach" in question:
    response = "Alexandra Oliveira and Catarina Santos teach this course."
elif "colab" in question:
    response = "We write our code in Google Colab."
else:
    response = "Sorry, I don't know that yet."

print("Response received")
````

### Step 4 — ✅ Display the answer

````python
print(f"FALC-ON AI: {response}")
````

### Step 5 — 🔁 Keep the conversation going

We put Steps 1 to 4 inside a `while True` loop. Typing `bye` leaves the loop with `break`, and a counter remembers how many questions were asked.

> 💡 Try: *"how many classes?"*, *"who teaches?"*, *"what is colab?"*, *"is there homework?"*, then *"bye"*.

````python
print("FALC-ON AI: Hi! Ask me about the course, or type bye to leave.")
question_count = 0

while True:
    # 👤 1. Receive the question
    question = input("You: ")
    if question == "bye":
        print(f"FALC-ON AI: Goodbye! You asked {question_count} question(s). Keep coding!")
        break
    question_count += 1

    # 🔧 2. Prepare the prompt - a real model will read it in Class 9
    prompt = f"You are a course assistant. Answer in one short sentence: {question}"

    # 🌐 3. Call the model - today, a stand-in made of if/elif rules
    if "class" in question:
        response = "The course has 10 classes."
    elif "teach" in question:
        response = "Alexandra Oliveira and Catarina Santos teach this course."
    elif "colab" in question:
        response = "We write our code in Google Colab."
    else:
        response = "Sorry, I don't know that yet."

    # ✅ 4. Display the answer
    print(f"FALC-ON AI: {response}")
````
