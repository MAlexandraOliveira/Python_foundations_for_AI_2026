## 1. Lists

In Class 1 you created lists and went through them with `for`. A **list** holds several values **in order**, in one variable.

### Reaching an item — by its position

Each item has a **position**, called its **index**. Python counts from **0**, not 1.

| `topics` | `"Python"` | `"AI"` | `"Data"` | `"Colab"` |
|----------|-----------|--------|----------|-----------|
| index | `0` | `1` | `2` | `3` |
| from the end | `-4` | `-3` | `-2` | `-1` |

````python
topics = ["Python", "AI", "Data", "Colab"]

print(topics)
print(len(topics))       # 4 - how many items
````

````python
print(topics[0])         # Python - the first item
print(topics[-1])        # Colab  - the last item
````

### Slicing — taking a piece of a list

`topics[1:3]` means *"from index 1, up to but **not including** 3"*. Leave out a number to go from the start, or to the end.

````python
print(topics[1:3])       # ['AI', 'Data'] - index 1 and 2, not 3
````

````python
print(topics[:2])        # ['Python', 'AI']   - from the start
print(topics[2:])        # ['Data', 'Colab']  - to the end
````

### Is this value in the list? — `in` and `not in`

In Class 1, `in` checked whether a word was inside a text. It works the same way on a list: is this value one of the items?

````python
print("AI" in topics)            # True
print("Excel" in topics)         # False
print("Excel" not in topics)     # True
````

### Adding an item — `.append()`

A list can change after you create it. `.append()` adds an item to the end.

`.append()` is a **method**: a tool that belongs to a value, written after it with a dot. (Section 2 shows how methods differ from built-in functions.)

````python
messages = ["Hi!", "What is AI?"]

messages.append("Thanks!")       # add to the end
print(messages)
````

### Replacing an item

Give the position, and the new value.

````python
messages[0] = "Hello!"           # replace the item at index 0
print(messages)
````

### Removing an item — `.remove()`

`.remove()` takes the **value** you want to remove.

````python
messages.remove("Thanks!")       # remove by value
print(messages)
````

### Counting a value — `.count()`

`.count()` tells you how many times a value appears in the list.

````python
ratings = [5, 3, 5, 4, 5, 2]

print(ratings.count(5))          # 3 - three ratings of 5
````

## 2. Built-in Functions for Lists

You already know some **built-in functions**: `print()`, `len()`, `round()`. They come with Python, and the value goes **inside the brackets**.

### `sum()`, `min()` and `max()`

The total, the smallest and the largest value of a list of numbers.

````python
print(sum(ratings))              # 24 - the total
print(min(ratings))              # 2 - the lowest rating
print(max(ratings))              # 5 - the highest rating
````

````python
# The average: the total divided by how many
average = sum(ratings) / len(ratings)
print(round(average, 1))         # 4.0
````

### `sorted()` — a sorted copy

`sorted()` gives you a **new** list, in order. The original list stays as it was.

````python
print(sorted(ratings))           # [2, 3, 4, 5, 5, 5]
print(ratings)                   # [5, 3, 5, 4, 5, 2] - unchanged
````

### Function or method?

You have now used both. They look alike — a name and brackets — but they are different kinds of tools:

| | Built-in function | Method |
|---|-------------------|--------|
| **How you write it** | the value goes **inside** the brackets: `len(ratings)` | the value comes **first**, then a dot: `ratings.count(5)` |
| **Who it belongs to** | Python itself — ready to use anywhere | one kind of value — list methods belong to lists |
| **Works on** | many kinds of values | only its own kind of value |
| **Changes the list?** | never — it gives you back a result | some do: `.append()` and `.remove()` change the list itself |
| **Examples** | `print()`, `len()`, `round()`, `sum()`, `sorted()` | `.append()`, `.remove()`, `.count()` |

> 💡 A built-in function is a tool on the workbench: you hand it a value, of almost any kind. A method is a button on the value itself: only values of that kind have it.

````python
# A built-in function works on many kinds of values
print(len(ratings))              # 6  - the items of a list
print(len("FALC-ON AI"))         # 10 - the characters of a text
````

A method only exists on its own kind of value: a list has `.append()`, but a text does not — `"hello".append("!")` stops with an error.

And the difference that matters most: a function gives you back a **new** result, while a method can **change** the value it belongs to.

````python
scores = [3, 1, 2]

sorted(scores)                   # a function gives back a new list...
print(scores)                    # [3, 1, 2] - ...scores is untouched

scores.append(4)                 # a method changes scores itself
print(scores)                    # [3, 1, 2, 4]
````

## 3. Sets — Unique Values

A **set** keeps only **unique** values, and has **no order**.

### Removing duplicates from a list — `set()`

`set(my_list)` builds a set from a list: every duplicate disappears.

To print the result, turn the set back into a list with `list()`.

> 💡 A set has no order, so the list can come out in **any order** — and the order can change from one run to the next.

````python
# The topic of each question users asked FALC-ON AI today
asked = ["class", "homework", "colab", "class", "exam", "homework"]

asked_topics = set(asked)
print(list(asked_topics))        # each topic once: class, colab, exam, homework - in any order
````

### In both sets — `.intersection()`

`a.intersection(b)` gives the values that are in **both** sets.

````python
# The topics FALC-ON AI knows how to answer
known_topics = set(["class", "teach", "colab"])

both = asked_topics.intersection(known_topics)    # asked and known
print(list(both))                                 # class, colab - in any order
````

### Only in the first set — `.difference()`

`a.difference(b)` gives the values that are in `a` but **not** in `b`.

````python
unknown = asked_topics.difference(known_topics)   # asked, but not known
print(list(unknown))                              # exam, homework - in any order
````

## 4. Dictionaries — Values with Names

This is the most important one.

A **dictionary** stores each value under a **name you choose**, called a **key**. Instead of remembering that position 2 is the language, you just ask for `"language"`.

It works like a real dictionary: you look up a **word** (the key) to find its **meaning** (the value).



- Curly brackets `{ }` hold the whole dictionary
- Each entry is a **key: value** pair — a colon `:` between them, a comma `,` between pairs
- You reach a value with square brackets and its key: `user["language"]`

````python
user = {"name": "Ana", "language": "Portuguese", "questions": 3}

print(user)
print(user["name"])              # Ana - by name, not by position
````

### Reading safely — `.get()`

Asking for a key that does not exist **stops your program** with an error.

`.get()` returns a fallback value instead: `user.get("email", "not given")`.

> ⚠️ Use `.get()` whenever you are not sure the key is there.

````python
print(user.get("email", "not given"))    # not given - no error
````

### Adding and updating

Assign to a key. If the key exists, its value is **updated**. If not, a new pair is **added**.

````python
user["questions"] = 4                    # update an existing key
user["email"] = "ana@example.com"        # add a new key
print(user)
````

### Starting from an empty dictionary

`{}` is an empty dictionary. You fill it as you go — very common inside a loop, for example to count things.

````python
counts = {}                      # empty
counts["python"] = 1             # a new key
counts["python"] += 1            # update it, like any variable
print(counts)                    # {'python': 2}
````

### Checking a key, and counting pairs — `in` and `len()`

`in` checks the **keys** of a dictionary. `len()` counts its pairs.

````python
print("name" in user)            # True
print("phone" in user)           # False
print(len(user))                 # 4 pairs
````

### Going through a dictionary — `for` and `.items()`

A `for` loop over a dictionary gives you its **keys**. `.items()` gives you the **key and the value together** — which is what you want almost every time.

````python
for key in user:
    print(key)
````

````python
for key, value in user.items():
    print(f"{key}: {value}")
````

### `.keys()` and `.values()`

`.keys()` gives the names and `.values()` gives the values. Wrap them in `list()` to print them as a list.

````python
print(list(user.keys()))         # ['name', 'language', 'questions', 'email']
print(list(user.values()))       # ['Ana', 'Portuguese', 4, 'ana@example.com']
````

## 6. AI Data — Messages and Conversations

### A message is a dictionary

When you talk to an AI model, each message is a small dictionary with two keys:

| Key | What it holds |
|-----|---------------|
| `"role"` | who is speaking: `"system"`, `"user"` or `"assistant"` (the model) |
| `"content"` | what they said |

The **system** message holds the instructions for the model — the prompt you prepared in the Class 1 capstone.

````python
message = {"role": "user", "content": "What is AI?"}

print(message["role"])
print(message["content"])
````

### A conversation is a list of messages

The whole conversation is a **list of message dictionaries**, in order. Every new message is appended to the end.

> 💡 A list of dictionaries is the most common shape in real data. It is exactly what a spreadsheet is: each dictionary is a row, each key is a column. Next class you will load one straight from a file with `pandas`.

````python
history = [
    {"role": "system", "content": "Answer briefly."},
    {"role": "user",   "content": "What is AI?"},
]

history.append({"role": "assistant", "content": "Software that learns from data."})
print(len(history))              # 3 messages
````

To go through the conversation, loop over the list: each `message` is one dictionary.

> 💡 Inside an f-string written with double quotes `" "`, write the keys with **single** quotes: `f"{message['role']}"`.

````python
for message in history:
    print(f"{message['role']}: {message['content']}")
````

### A request to the model — a dictionary holding the conversation

To call a model, the app sends one dictionary: which **model** to use, and the **messages**. A dictionary holding a list, holding dictionaries:



To reach a value deep inside, name each step, left to right: `request["messages"][1]["content"]` is *"in request, the messages, the second one, its content"*.

````python
request = {
    "model": "gemini",
    "messages": [
        {"role": "system", "content": "Answer briefly."},
        {"role": "user",   "content": "What is AI?"},
    ],
}

print(request["model"])                        # gemini
print(request["messages"][1]["content"])       # What is AI?
````

## 7. JSON — How AI Data Travels

A Python dictionary only exists inside your program. To send it to an AI model over the internet, it has to become **text**.
**JSON** is the text format almost every API uses — and it looks almost exactly like Python dictionaries and lists.

Python comes with a **library** (a ready-made toolbox) called `json`. `import json` makes it available.

| Python | JSON |
|--------|------|
| `dict` | `{ }` object |
| `list` | `[ ]` array |
| `True` / `False` | `true` / `false` |

````python
import json
print("json is ready")
````

### `json.dumps()` — Python to JSON text

`json.dumps()` turns a dictionary or a list into JSON text. `indent=2` makes it easy to read.

````python
text = json.dumps(request, indent=2)

print(type(text))                # type: str - it is now text
print(text)
````

> 🤖 **AI Connection**
>
> Every call to an AI model works like this: build a dictionary, turn it into JSON text, send it. The model's reply comes back as JSON text too, and Python turns it back into a dictionary. In Class 9 you will make that call for real.

## 8. 🔗 Capstone — FALC-ON AI Gets a Memory

In Class 1, FALC-ON AI's stand-in model was a chain of `if` / `elif` rules, and it forgot every question straight away.
Today it gets two upgrades:

1. Its **knowledge** becomes a dictionary: each keyword points to its answer. Adding a topic is adding one line.
2. It gets a **memory**: the conversation is kept as a list of messages — the same shape an AI model receives — and saved as JSON at the end.

### Step 1 — The knowledge, as a dictionary

````python
knowledge = {
    "class": "The course has 10 classes.",
    "teach": "Alexandra Oliveira and Catarina Santos teach this course.",
    "colab": "We write our code in Google Colab.",
}

print(knowledge["colab"])
````

### Step 2 — Find the answer to one question

Go through the keywords. If one is in the question, take its answer.

````python
question = "who teaches this course?"

response = "Sorry, I don't know that yet."
for keyword in knowledge:
    if keyword in question:
        response = knowledge[keyword]

print(response)
````

### Step 3 — The conversation, with a memory

The conversation starts with the system message. Each question and each answer is appended to `history`.

> 💡 Try: *"how many classes?"*, *"who teaches?"*, *"is there homework?"*, then *"bye"*. Type in lowercase.

````python
history = [
    {"role": "system", "content": "You are a course assistant. Answer in one short sentence."}
]

print("FALC-ON AI: Hi! Ask me about the course, or type bye to leave.")

while True:
    # 👤 1. Receive the question, and remember it
    question = input("You: ")
    if question == "bye":
        print("FALC-ON AI: Goodbye! Keep coding!")
        break
    history.append({"role": "user", "content": question})

    # 🌐 2. Call the model - today, a look-up in the knowledge dictionary
    response = "Sorry, I don't know that yet."
    for keyword in knowledge:
        if keyword in question:
            response = knowledge[keyword]

    # ✅ 3. Display the answer, and remember it
    history.append({"role": "assistant", "content": response})
    print(f"FALC-ON AI: {response}")
````

### Step 4 — Save the conversation as JSON

This is the text an app would send to a real model, or store to read later.

`ensure_ascii=False` keeps accented letters readable: without it, a word like "olá" would be saved as "ol\u00e1".

````python
print(f"Messages in the conversation: {len(history)}")
print(json.dumps(history, indent=2, ensure_ascii=False))
````
