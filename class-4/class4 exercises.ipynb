### Exercise D2 — FALC-ON AI with more knowledge

The cell below gives you `make_message()`, `clean_text()` and the stand-in model `find_answer()` from the capstone.

1. Add at least **two** new keywords to `knowledge`: `"hours"` → `"Each class lasts 3 hours."` and `"project"` → `"The final project is in Class 10."`
2. Write `chat_turn(question, history)`: it cleans the question, finds the answer, appends **both** messages to `history`, and returns the answer
3. Start a `history` with one system message, `"Answer briefly."`
4. For each question in `questions`, print `You: <question>` and `FALC-ON AI: <answer>`
5. At the end, print how many messages the history holds

Expected output:
```
You: How many HOURS is each lesson?
FALC-ON AI: Each class lasts 3 hours.
You:   when is the PROJECT? 
FALC-ON AI: The final project is in Class 10.
You: Who teaches?
FALC-ON AI: Alexandra Oliveira and Catarina Santos teach this course.
Messages in the history: 7
```

````python
# D2
def make_message(role, content):
    """Return one chat message."""
    return {"role": role, "content": content}

def clean_text(text):
    """Return the text in lower case, without spaces at the ends."""
    return text.lower().strip()

knowledge = {
    "class": "The course has 10 classes.",
    "teach": "Alexandra Oliveira and Catarina Santos teach this course.",
}

def find_answer(question):
    """The stand-in model: answer from the knowledge dictionary."""
    for keyword in knowledge:
        if keyword in question:
            return knowledge[keyword]
    return "Sorry, I don't know that yet."

questions = ["How many HOURS is each lesson?", "  when is the PROJECT? ", "Who teaches?"]
````
