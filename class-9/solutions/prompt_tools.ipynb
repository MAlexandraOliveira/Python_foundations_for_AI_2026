- a **module docstring** on the first line — `"""Helpers to prepare prompts for an AI model."""`
- a value `MAX_CHARS = 200`
- `clean_text(text)` — returns the text in lower case, without spaces at the ends
- `is_too_long(text)` — returns `True` when the text has more than `MAX_CHARS` characters

> ⚠️ Keep this notebook **in the same folder** as the exercises notebook.

````python
"""Helpers to prepare prompts for an AI model."""

MAX_CHARS = 200


def clean_text(text):
    """Return the text in lower case, without spaces at the ends."""
    return text.lower().strip()


def is_too_long(text):
    """True when the text has more than MAX_CHARS characters."""
    return len(text) > MAX_CHARS


def make_prompt(question, facts):
    """Build a prompt that asks the model to answer from the facts only."""
    return f"Using only these facts: {facts}, answer: {question}"
````
